# Analisis Pariwisata Indonesia 2026

Notebook pendamping proyek portofolio. Data dikompilasi dari publikasi resmi BPS (lihat `data/SOURCES.md`). Jalankan sel dari atas ke bawah untuk mereproduksi angka-angka pada `findings.json`.

```python
import pandas as pd

tahunan = pd.read_csv('data/wisman_tahunan.csv')
bulanan = pd.read_csv('data/wisman_bulanan.csv')
pintu   = pd.read_csv('data/wisman_pintu_masuk.csv')
keb     = pd.read_csv('data/wisman_kebangsaan.csv')
tpk     = pd.read_csv('data/tpk_provinsi.csv')
wisnus  = pd.read_csv('data/wisnus_provinsi.csv')
tahunan
```

## 1. Kolaps & pemulihan terhadap baseline 2019

```python
base = tahunan.loc[tahunan.tahun == 2019, 'kunjungan_wisman'].iloc[0]
penuh = tahunan[tahunan.status_periode.str.startswith('Setahun penuh')].copy()
penuh['pemulihan_persen'] = (100 * penuh.kunjungan_wisman / base).round(1)
penuh[['tahun','kunjungan_wisman','pemulihan_persen']]
# Titik terendah 2021 = 9,7%; 2025 = 95,5%
```

## 2. Konsistensi bulanan 2026 vs kumulatif resmi BPS (10.578.286)

```python
assert bulanan[bulanan.tahun == 2026].kunjungan_wisman.sum() == 10578286
bulanan[bulanan.tahun == 2026].sort_values('bulan_ke')
# Puncak: Agustus 1.599.660
```

## 3. Konsentrasi pintu masuk 2025

```python
p25 = pintu[pintu.periode.astype(str) == '2025'].copy()
p25['pangsa_persen'] = (100 * p25.kunjungan / 15386646).round(2)
p25.sort_values('kunjungan', ascending=False)
# Ngurah Rai = 44,89%; dua gerbang teratas = 62,8%
```

## 4. Pemulihan kebangsaan 2023 vs 2019

```python
piv = keb[keb.periode.isin(['2019','2023'])].pivot(index='kebangsaan', columns='periode', values='kunjungan')
piv['pemulihan_persen'] = (100 * piv['2023'] / piv['2019']).round(1)
piv.sort_values('pemulihan_persen', ascending=False)
# Australia 103,2% tertinggi; Tiongkok 38,0% terendah
```

## 5. TPK provinsi dan wisnus

```python
print(tpk[tpk.provinsi != 'INDONESIA'].nlargest(5, 'tpk_bintang_tahunan_2025_persen')[['provinsi','tpk_bintang_tahunan_2025_persen']])
print(wisnus[wisnus.provinsi_tujuan != 'INDONESIA'].nlargest(5, 'perjalanan_2025'))
# Wisnus 2025 total 1.200.332.559 = 78x wisman 2025
```

## Kesimpulan

- Pemulihan wisman 2025 = 95,5% dari 2019; Jan–Agu 2026 = 98,7% dari periode sama 2019.
- Konsentrasi gerbang ekstrem: Ngurah Rai 44,9% (2025).
- TPK bintang nasional turun ke 49,30% meski kunjungan naik.
- Wisnus 1,20 miliar perjalanan = 78 kali wisman; 64,8% bertujuan ke Jawa.
